#  Section 1: Data Cleaning

```python
import pandas as pd
import numpy as np
df = pd.read_csv('health_and_wellness_cleaned_20230810_213000.csv')
pd.options.display.float_format = '{:,.2f}'.format
```

```python
df.drop_duplicates(inplace=True)
```

```python
df
```

```python
df.dtypes
```

```python
df['Price']
```

```python
df['Price'] = df['Price'].astype(str).str.replace(r'[^\d.]', '', regex=True).astype(float)
```

```python
df.isnull().sum()
```

```python
df
```

```python
df['Shop Location'].value_counts()
```

```python
df[df['Shop Location'].isnull()][['Name','Price','Total Sold','Total Reviews']]
```

```python
condition = (df['Total Sold'] == 0) & (df['Total Reviews'] == 0) & (df['Shop Location'].isnull())
df = df[~condition].copy()
```

```python
df['Shop Location'] = df['Shop Location'].fillna("Unknown")
```

```python
df['Total Sold'] = df['Total Sold'].str.replace(r'[^\d]', '', regex=True)
df['Total Reviews'] = df['Total Reviews'].str.replace(r'[^\d]', '', regex=True)
```

```python
df['Total Sold'] = df['Total Sold'].replace('', None).fillna(0).astype(int)
df['Total Reviews'] = df['Total Reviews'].replace('',None).fillna(0).astype(int)
```

```python
df
```

```python
df['Revenue'] = df['Total Sold'] * df['Price']
```

```python
df['Review Rate'] = df['Total Reviews'] / (df['Total Sold'] + 1)
```

```python
section_analysis = df.groupby('Section').agg(Product_count=('Price', 'count'),
                                             Avg_price=('Price','mean'),
                                             Avg_sold=('Total Sold','mean'),
                                             Avg_reviews=('Total Reviews','mean'),
                                             Total_Revenue=('Revenue','sum'),
                                             Avg_Review_rate=('Review Rate','mean')).round(2).sort_values('Total_Revenue',ascending=False)
section_analysis
```

```python
df.groupby('Shop Location').agg(Product_Count=('Price', 'count'),
    Avg_Price=('Price', 'mean'),
    Avg_Sold=('Total Sold', 'mean'),
    Total_Revenue=('Revenue', 'sum')).round(2).sort_values('Total_Revenue', ascending=False).head(15)
```

```python
#df['Price Segment'] = pd.cut(df['Price'],
                             #bins=[0,200,500,1000,6000],
                             #labels=['Budget', 'Mid-Range', 'Premium', 'Luxury'])
condition = [df['Price'] <= 200,
    df['Price'] <= 500,
    df['Price'] <= 1000,
    df['Price'] > 1000]
choices = ['Budget', 'Mid-Range', 'Premium', 'Luxury']
df['Price Segment'] = np.select(condition,choices,default='')

df.groupby('Price Segment').agg(
    Count=('Price', 'count'),
    Avg_Sold=('Total Sold', 'mean'),
    Avg_Reviews=('Total Reviews', 'mean'),
    Total_Revenue=('Revenue', 'sum')
).round(2)
```

```python
df[['Price', 'Total Sold', 'Total Reviews', 'Revenue']].corr().round(2)
```

```python
df['Location Type'] = df['Shop Location'].apply(lambda x:'Bangkok' if x == 'Bangkok' else 'Rural')

df.groupby('Location Type').agg(Count=('Price','count'),
                                Avg_price=('Price','mean'),
                                Avg_sold=('Total Sold', 'mean'),
                                Avg_reviews=('Total Reviews', 'mean'),
                                Total_Revenue=('Revenue','sum')).round(2)
```

#  Section 4: Machine Learning

```python
df
```

```python
x = df[['Total Sold', 'Total Reviews', 'Review Rate', 'Section', 'Location Type']].copy()
y = df['Price']

x = pd.get_dummies(x,columns=['Section', 'Location Type'])
x
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
print(X_train.shape)
print(X_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn import set_config
set_config(display='text')
lm = LinearRegression()
lm.fit(X_train,y_train)
lm.score(X_train,y_train)
```

```python
from sklearn.ensemble import RandomForestRegressor
rf = RandomForestRegressor()
rf.fit(X_train,y_train)
```

```python
print('Linear Regression R²:', lm.score(X_test, y_test))
print('Random Forest R²:', rf.score(X_test, y_test))
```

```python
x = df[['Price', 'Total Reviews', 'Review Rate', 'Section', 'Location Type']].copy()
y = df['Total Sold']

x = pd.get_dummies(x, columns=['Section', 'Location Type'])

from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

lm2 = LinearRegression()
lm2.fit(X_train,y_train)

rf2 = RandomForestRegressor()
rf2.fit(X_train,y_train)

print('Linear R²:', lm2.score(X_test, y_test))
print('Random Forest R²:', rf2.score(X_test, y_test))
```

```python
importance = pd.DataFrame({'Feature' : X_train.columns ,
                           'Importance' : rf2.feature_importances_}).sort_values('Importance',ascending=False)
importance.head(10)
```

```python
x2 = df[['Price', 'Review Rate', 'Section', 'Location Type']].copy()
y2 = df['Total Sold']

x2 = pd.get_dummies(x2, columns=['Section', 'Location Type'])

X_train2, X_test2, y_train2, y_test2 = train_test_split(x2, y2, test_size=0.2, random_state=42)

rf3 = RandomForestRegressor(n_estimators=100, random_state=42)
rf3.fit(X_train2, y_train2)

print('Random Forest R² (no leakage):', rf3.score(X_test2, y_test2))
```

```python
importance2 = pd.DataFrame({
    'Feature': X_train2.columns,
    'Importance': rf3.feature_importances_
}).sort_values('Importance', ascending=False)

importance2.head(10)
```
